```python

%%writefile dog_info.py
"""
This module creates and formats dog-care records.
"""

def create_dog_record(name, age, feeding, notes, date_recorded):
    """Validate and organize dog-care information."""
    if not name:
        raise ValueError("The dog's name cannot be blank.")
    if age < 0:
        raise ValueError("Age cannot be negative.")

    return {
        "name": name,
        "age": age,
        "feeding": feeding or "No feeding instructions provided",
        "notes": notes or "No additional notes",
        "date_recorded": date_recorded
    }


def format_dog_record(record):
    """Format a dog-care record for display."""
    return (
        f"Name: {record['name']}\n"
        f"Age: {record['age']} years\n"
        f"Feeding: {record['feeding']}\n"
        f"Notes: {record['notes']}\n"
        f"Date recorded: {record['date_recorded']}\n"
        "------------------"
    )
```

```python

%%writefile file_handler.py
"""
This module saves and loads dog-care records using CSV files.
"""

import csv
import os

FILE_NAME = "dog_records.csv"
FIELDNAMES = ["name", "age", "feeding", "notes", "date_recorded"]


def save_dog_record(record):
    """Save a dog-care record to a CSV file."""
    file_exists = os.path.exists(FILE_NAME)

    with open(FILE_NAME, "a", newline="", encoding="utf-8") as file:
        writer = csv.DictWriter(file, fieldnames=FIELDNAMES)

        if not file_exists or os.path.getsize(FILE_NAME) == 0:
            writer.writeheader()

        writer.writerow(record)


def load_dog_records():
    """Load saved dog-care records from the CSV file."""
    if not os.path.exists(FILE_NAME):
        return []

    with open(FILE_NAME, "r", newline="", encoding="utf-8") as file:
        records = list(csv.DictReader(file))

    for record in records:
        record["age"] = int(record["age"])

    return records
```

```python


import datetime
import dog_info
from file_handler import save_dog_record, load_dog_records


def main():
    """Run the Dog Care Organizer menu."""
    while True:
        print("\nDog Care Organizer")
        print("1. Add a dog-care record")
        print("2. View saved records")
        print("3. Exit")

        choice = input("Choose an option: ").strip()

        if choice == "1":
            name = input("Dog's name: ").strip()

            try:
                age = int(input("Dog's age in years: "))
            except ValueError:
                print("Please enter a valid whole number for age.")
                continue

            feeding = input("Feeding instructions: ").strip()
            notes = input("Additional care notes: ").strip()

            try:
                record = dog_info.create_dog_record(
                    name,
                    age,
                    feeding,
                    notes,
                    datetime.date.today().isoformat()
                )

                save_dog_record(record)
                print("Record saved successfully!")

            except ValueError as error:
                print(error)

        elif choice == "2":
            records = load_dog_records()

            if not records:
                print("No records have been saved yet.")
            else:
                for record in records:
                    print(dog_info.format_dog_record(record))

        elif choice == "3":
            print("Goodbye!")
            break

        else:
            print("Please choose 1, 2, or 3.")


main()
```
