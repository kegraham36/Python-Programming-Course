```python

# Assignment 4 - Functions and Error Handling

APP_NAME = "Simple Calculator"


def validate_number(value):
    """Convert a value to a number and reject invalid input."""
    try:
        return float(value)
    except ValueError:
        raise ValueError("Please enter a valid number.")


def calculate(first, second, operation):
    """Perform a math operation and return the result."""
    if operation == "+":
        return first + second
    elif operation == "-":
        return first - second
    elif operation == "*":
        return first * second
    elif operation == "/":
        if second == 0:
            raise ZeroDivisionError("Cannot divide by zero.")
        return first / second
    else:
        raise ValueError("Please choose +, -, * or /.")


def format_result(result, decimal_places=2):
    """Round a result to the requested number of decimal places."""
    return round(result, decimal_places)


def run_calculator():
    """Get user input, calculate the answer, and handle errors."""
    print(APP_NAME)  # APP_NAME is a global variable.

    try:
        first = validate_number(input("Enter the first number: "))
        operation = input("Choose an operation (+, -, *, /): ")
        second = validate_number(input("Enter the second number: "))

        result = calculate(first, second, operation)
        # result is a local variable inside this function.

        print("Result:", format_result(result))

    except ValueError as error:
        print("Input error:", error)
    except ZeroDivisionError as error:
        print("Math error:", error)


# Run the calculator
run_calculator()

# Test the functions with different arguments
print("\nFunction tests:")
print("Addition:", calculate(5, 3, "+"))
print("Division:", calculate(10, 2, "/"))
print("Rounded result:", format_result(3.14159))
print("More precise result:", format_result(3.14159, 4))
```
