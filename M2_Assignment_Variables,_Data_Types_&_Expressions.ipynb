```python
# Assignment 2 - Variables and Expressions

# Numeric variables
hours_worked = 8
hourly_rate = 18
days_worked = 5

weekly_hours = hours_worked * days_worked
weekly_pay = weekly_hours * hourly_rate

print("I worked", weekly_hours, "hours this week.")
print("My estimated weekly pay is $", weekly_pay)

# String variable and manipulation
first_name = "Kailyn"
last_name = "Graham"

full_name = first_name + " " + last_name
name_part = full_name[:6]

print("My full name is " + full_name + ".")
print("The first part of my name is " + name_part + ".")

# Boolean expression
more_than_30_hours = weekly_hours > 30

print("I worked more than 30 hours this week:", more_than_30_hours)
```
